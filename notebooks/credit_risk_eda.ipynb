### Age-Based Delinquency Analysis — Preliminary Finding

**Business question:** How does the serious delinquency rate vary across age groups?

The exploratory analysis shows a declining serious delinquency rate across the defined age groups. The rate was highest among borrowers under 30 at 11.73%, compared with 10.07% for ages 30–39, 8.37% for ages 40–49, 6.45% for ages 50–59, and 3.10% for borrowers aged 60 and above.

**Initial interpretation:** Younger borrowers in this dataset have a higher observed incidence of serious delinquency. Potential explanations include differences in income stability, credit history, debt obligations, or financial resilience. These are hypotheses, not established causes.

**Limitations:** This is a descriptive comparison and does not establish causation. Earlier data-quality checks identified unusual age values, which have not been corrected or excluded. Further analysis of income, debt ratios, and other relevant variables is needed before drawing stronger conclusions.

**Status:** Preliminary finding — subject to refinement after further analysis.

**1. Overall delinquency**

Higher recorded past-due counts were generally associated with higher serious delinquency rates in the common count groups. Some higher-count groups were small, and the meaning of unusual values such as 98 remains unverified.

These are descriptive findings from the available dataset. They do not establish causation, validate a lending policy, or demonstrate how a future portfolio will perform. ```python

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Locate the project and load the dataset
project_folder = Path("..").resolve()
data_path = project_folder / "data" / "cs-training.csv"
charts_folder = project_folder / "outputs" / "charts"
charts_folder.mkdir(parents=True, exist_ok=True)

df_chart = pd.read_csv(data_path)

late_col = "NumberOfTimes90DaysLate"
target_col = "SeriousDlqin2yrs"

# Exclude missing values for these two variables only
plot_data = df_chart[[late_col, target_col]].dropna().copy()

# Group recorded counts without hiding unusually high values
def late_payment_group(value):
    if value == 0:
        return "0"
    elif value == 1:
        return "1"
    elif value == 2:
        return "2"
    elif value == 3:
        return "3"
    elif value <= 10:
        return "4–10"
    else:
        return "11+ (review)"

plot_data["LatePaymentGroup"] = (
    plot_data[late_col].apply(late_payment_group)
)

group_order = ["0", "1", "2", "3", "4–10", "11+ (review)"]

summary = (
    plot_data.groupby("LatePaymentGroup", observed=False)[target_col]
    .agg(["count", "mean"])
    .reindex(group_order)
    .dropna(subset=["count"])
)

summary["DelinquencyRatePct"] = summary["mean"] * 100

# Create and save the chart
fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(
    summary.index,
    summary["DelinquencyRatePct"]
)

ax.set_title("Serious Delinquency by 90-Day Late-Payment History")
ax.set_xlabel("Recorded Number of 90-Day Late Payments")
ax.set_ylabel("Observed Serious Delinquency Rate (%)")
ax.grid(axis="y", alpha=0.25)
ax.set_axisbelow(True)

for bar, rate in zip(bars, summary["DelinquencyRatePct"]):
    ax.annotate(
        f"{rate:.1f}%",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        ha="center",
        va="bottom",
        xytext=(0, 4),
        textcoords="offset points"
    )

fig.tight_layout()

chart_path = (
    charts_folder / "serious_delinquency_by_90day_lates.png"
)
fig.savefig(chart_path, dpi=150, bbox_inches="tight")
plt.show()
plt.close(fig)

print("Chart saved to:", chart_path)
print("Groups used in the chart:")
print(summary[["count", "DelinquencyRatePct"]].round(2))
```
